# Stage 1 road construction (Card 2.3)

Builds each Stage 1 lane with dream-gym's `Road`, following the steps of dream-gym's `ad_gym_road_construction.ipynb` and its "Small-Scale Cone Track" road example. The road specs come from `road_spec()` in `stage1_scenarios.py`, so the test cases, this notebook and the floor drawings in `docs/figures/` always use the same lanes.

Run from `tests/policy/` with dream-gym installed (`pip install wheels/*.whl`).

```python
import matplotlib.pyplot as plt
from dreamgym.envs import Road

import stage1_scenarios as s1
from stage1_tuning_cases import TUNING_CASES
```

## The road spec of the main Stage 1 lane

One straight element with the cones, then a run-out without cones so the simulated car can overrun the stop zone.

```python
import pprint
pprint.pprint(s1.road_spec(s1.STAGE1_LANE), sort_dicts=False)
```

## Every tuning lane

Change a `Lane` here to try another size, then add a case using it to `stage1_tuning_cases.py`.

```python
lanes = list(dict.fromkeys(case.lane for case in TUNING_CASES))
fig, axes = plt.subplots(len(lanes), 1, figsize=(10, 2.2 * len(lanes)), sharex=True)
for axis, lane in zip(axes, lanes):
    road = Road(road_spec=s1.road_spec(lane))
    road.plot_road_on_matplotlib_axis(axis, progress_step_m=0.05)
    axis.set_aspect("equal")
    axis.set_title(f"Lane {lane.name}: {lane.cones_per_row} cones per row")
axes[-1].set_xlabel("world x [m]")
fig.tight_layout()
plt.show()
```

## Check the cones against the drawing

```python
for lane in lanes:
    xy, colour_ids = Road(road_spec=s1.road_spec(lane)).get_cone_arrays_from_spec()
    simulated = sorted((int(c), round(float(x), 6), round(float(y), 6)) for (x, y), c in zip(xy, colour_ids))
    drawn = sorted((c.value, round(x, 6), round(y, 6)) for c, _, x, y in s1.all_cones(lane))
    print(lane.name, "matches the drawing" if simulated == drawn else "DOES NOT match the drawing")
```
